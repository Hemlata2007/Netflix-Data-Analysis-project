```python
import pandas as pd

# डेटा लोड करें
df = pd.read_csv('/Dataset.csv')

# डेटा की पहली 5 पंक्तियाँ देखें
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()

df['type'] = df['type'].str.strip().str.title()
df['rating'] = df['rating'].str.strip().str.upper()


df.to_csv('netflix_cleaned.csv', index=False)

print("Task 1 Completed! Cleaned file saved as 'netflix_cleaned.csv'")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = pd.read_csv('netflix_cleaned.csv')

content_counts = df_clean['type'].value_counts()
print("Content Distribution:")
print(content_counts)


plt.figure(figsize=(6,4))
sns.countplot(x='type', data=df_clean, palette='pastel')
plt.title('Distribution of Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')

plt.show()
```

```python
top_countries = df_clean['country'].value_counts().head(10)
print("Top 10 Content Producing Countries:")
print(top_countries)

plt.figure(figsize=(10,5))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Content Producing Countries on Netflix')
plt.xlabel('Number of Movies / TV Shows')
plt.ylabel('Country')


plt.show()
```

```python
yearly_trends = df_clean['release_year'].value_counts().sort_index()

yearly_trends = yearly_trends[yearly_trends.index >= 2000]


plt.figure(figsize=(12,6))
plt.plot(yearly_trends.index, yearly_trends.values, marker='o', color='red', linewidth=2)
plt.title('Netflix Content Release Trends (2000 - Present)')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies / TV Shows')
plt.grid(True, linestyle='--', alpha=0.6)


plt.show()
```

```python
top_ratings = df_clean['rating'].value_counts().head(10)
print("Top 10 Content Ratings:")
print(top_ratings)

plt.figure(figsize=(10,5))
sns.barplot(x=top_ratings.index, y=top_ratings.values, palette='magma')
plt.title('Top 10 Content Ratings on Netflix')
plt.xlabel('Rating Category')
plt.ylabel('Count')


plt.show()
```

```python
pivot_df = pd.crosstab(df_clean['release_year'], df_clean['type'])
pivot_df = pivot_df[pivot_df.index >= 2010]


plt.figure(figsize=(12,6))
plt.plot(pivot_df.index, pivot_df['Movie'], marker='o', color='blue', label='Movies', linewidth=2)
plt.plot(pivot_df.index, pivot_df['TV Show'], marker='s', color='orange', label='TV Shows', linewidth=2)
plt.title('Netflix Business Growth: Movies vs TV Shows (2010 - Present)')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)


plt.show()

print("\n--- FINAL BUSINESS INSIGHTS ---")
print("1. Netflix contains significantly more Movies than TV Shows across the platform.")
print("2. The United States and India are the primary content hubs powering Netflix's library.")
print("3. Content production peaked rapidly between 2017-2019, heavily driven by Adult ('TV-MA') rated content.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df_filtered = df_clean[df_clean.iloc[:, 6] >= 2010]


movies_per_year = df_filtered[df_filtered.iloc[:, 1].str.contains('Movie', case=False, na=False)].iloc[:, 6].value_counts().sort_index()
shows_per_year = df_filtered[df_filtered.iloc[:, 1].str.contains('Show', case=False, na=False)].iloc[:, 6].value_counts().sort_index()


plt.figure(figsize=(12, 6))
plt.plot(movies_per_year.index, movies_per_year.values, marker='o', color='blue', label='Movies', linewidth=2)
plt.plot(shows_per_year.index, shows_per_year.values, marker='s', color='orange', label='TV Shows', linewidth=2)

plt.title('Netflix Business Growth: Movies vs TV Shows (2010 - Present)')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)


plt.show()


print("\n--- FINAL BUSINESS INSIGHTS ---")
print("1. Netflix contains significantly more Movies than TV Shows across the platform.")
print("2. The United States and India are the primary content hubs powering Netflix's library.")
print("3. Content production peaked rapidly between 2017-2019, heavily driven by Adult rated content.")
```
